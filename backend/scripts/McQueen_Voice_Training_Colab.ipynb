```python
# Step 2: Upload your McQueen Voice Reference WAV file
from google.colab import files
import os

print("🎙️ Please upload your clean 'mcqueen_ref.wav' file (15-60 seconds of audio):")
uploaded = files.upload()
ref_filename = list(uploaded.keys())[0]
print(f"✅ Uploaded: {ref_filename}")
```

```python
# Step 3: Run Zero-Shot Voice Synthesis with F5-TTS / XTTS v2
from TTS.api import TTS
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"🚀 Running on device: {device}")

# Load XTTS v2 Multi-Lingual & Cross-Lingual Model
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(device)

# Synthesize Test Speech
test_text = "Ka-chow! I'm Emiot, and I'm ready to race! What are we building today?"
output_wav = "emiot_mcqueen_test.wav"

tts.tts_to_file(
    text=test_text,
    speaker_wav=ref_filename,
    language="en",
    file_path=output_wav
)

print(f"🎉 Generated sample audio saved to '{output_wav}'!")
```
